```python
import requests 
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = 'https://egypt-store.net/'
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36',
    'Accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8',
    'Accept-Language': 'en-US,en;q=0.5',
    'Accept-Encoding': 'gzip, deflate',
    'Connection': 'keep-alive',
}
```

```python
request = requests.get(url, headers=headers)
request.encoding = "utf-8"
print(request)
soup = BeautifulSoup(request.content, 'html')

```

```python
    #cleaner verstion of the code 
priceslist= []

for price in soup.find_all('span', class_='price'): 
    del_tag = price.find('del')
    ins_tag = price.find('ins')


    if del_tag is None and ins_tag is None:
        range_amount = []
        for bdi in price.find_all('bdi'): 
            range_amount.append(bdi.get_text())
        priceslist.append({ 'Price_range' :'-'.join(range_amount) , 'discounted_Price' : None})

    else:
        priceslist.append({'original_price' : del_tag.find('bdi').get_text(), 'discounted_Price' : ins_tag.find('bdi').get_text()})
print(priceslist)
```
