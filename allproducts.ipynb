```python
from bs4 import BeautifulSoup 
import requests
import pandas as pd 
```

```python
url = 'https://egypt-store.net/shop/'
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36',
    'Accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8',
    'Accept-Language': 'en-US,en;q=0.5',
    'Accept-Encoding': 'gzip, deflate',
    'Connection': 'keep-alive',
}

```

```python
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.content, 'html.parser')
print(soup.prettify())
```

```python
page_numbers = soup.find('ul', class_="page-numbers")
links = []
for num in page_numbers.find_all('li'):
   if num.get_text() != '→':
        pg_num = num.get_text()
        links.append('https://egypt-store.net/shop/page/' + pg_num + '/')
        print(links)
```

```python
soup.find_all('div', class_ = 'product-wrapper')[0].find('del').get_text()
```

```python
data = []
product = soup.find_all('div', class_ = 'product-wrapper')
for item in product:
    title = item.find('h3',class_="wd-entities-title" ).get_text().strip()
    category = item.find('div', class_ = 'wd-product-cats').get_text().strip()
    original_price = item.find('del')
    price = item.find('ins')
    if original_price is None and price is None: 
        range_amount = []
        for bdi in item.find_all('bdi'): 
            range_amount.append(bdi.get_text())
        data.append({'Title' : title, 'Category' : category, 'Original_Price':  original_price ,  'Price' :'-'.join(range_amount) })
        print(data)    
       
    else: 
        data.append({'Title' : title, 'Category' : category, 'Original_Price':  original_price.get_text() , "Price" : price.get_text() })
        print( data)
    
```

```python
def get_page_data (link) : 
    url = link
    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.content, 'html.parser')
    product = soup.find_all('div', class_ = 'product-wrapper')
    for item in product:
        title = item.find('h3',class_="wd-entities-title" ).get_text().strip()
        category = item.find('div', class_ = 'wd-product-cats').get_text().strip()
        original_price = item.find('del')
        price = item.find('ins')
        if original_price is None and price is None: 
            range_amount = []
            for bdi in item.find_all('bdi'): 
                range_amount.append(bdi.get_text())
            data.append({'Title' : title, 'Category' : category, 'Original_Price':  original_price ,  'Price' :'-'.join(range_amount) })
            print(data)    
        
        else: 
            data.append({'Title' : title, 'Category' : category, 'Original_Price':  original_price.get_text() , "Price" : price.get_text() })
            print( data)
```

```python
for link_ in links : 
    get_page_data(link_)
```

```python
data_frame = pd.DataFrame(data)
data_frame.to_csv('All_products.csv', index=False ,  encoding="utf-8-sig")
```
